```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats # Import added here

# Configuração visual
sns.set_theme(style="whitegrid")
plt.rcParams["font.sans-serif"] = "DejaVu Sans"
Azul="#0b1f3b"
Amarelo="#ffc857"

# Fixar semente para reprodutibilidade
np.random.seed(42)

# Dados de qualificação do Bahrein 2022
bahrain_2022_q1_times = np.sort([
    convert_to_seconds(d['q1']) for d in data_bahrain_2022 if d['q1'] is not None
])

# Dados de qualificação do Bahrein 2023
bahrain_2023_q1_times_filtered = np.sort([t for t in bahrain_2023_q1_times if not np.isnan(t)])

# Para garantir que ambos os arrays tenham o mesmo número de elementos para o QQ-Plot, vamos truncar
# o array mais longo para o tamanho do mais curto.
min_len_bahrain = min(len(bahrain_2022_q1_times), len(bahrain_2023_q1_times_filtered))

bahrain_2022_q1_times_plot = bahrain_2022_q1_times[:min_len_bahrain]
bahrain_2023_q1_times_plot = bahrain_2023_q1_times_filtered[:min_len_bahrain]

# DataFrame estruturado
df = pd.DataFrame({
    'Piloto_Rank': range(1, min_len_bahrain + 1),
    '2022': bahrain_2022_q1_times_plot,
    '2023': bahrain_2023_q1_times_plot
})

df_melted = pd.melt(df, id_vars=['Piloto_Rank'], value_vars=['2022', '2023'],
                    var_name='Ano', value_name='Tempo_s')

# Criando o painel de 3 gráficos
fig, axes = plt.subplots(1, 3, figsize=(18, 5.5), dpi=100)

# -----------------------------------------------------------------------------
# PASSO 1: GRÁFICO DE BARRAS (MÉDIA)
# -----------------------------------------------------------------------------
# Calculate statistics for mean and confidence interval
stats_df = df_melted.groupby('Ano')['Tempo_s'].agg(
    mean='mean',
    std='std',
    count='count'
).reset_index()

# Calculate Standard Error of the Mean (SEM)
stats_df['sem'] = stats_df['std'] / np.sqrt(stats_df['count'])

# Calculate 95% Confidence Interval for the mean using t-distribution
stats_df['yerr'] = stats_df.apply(
    lambda row: row['mean'] - stats.t.interval(0.95, df=row['count']-1, loc=row['mean'], scale=row['sem'])[0]
    if row['count'] > 1 else 0,
    axis=1
)

bars = axes[0].bar(stats_df['Ano'], stats_df['mean'], color=[Azul, Amarelo],
                   width=0.45, alpha = 0.7, edgecolor='black', yerr=stats_df['yerr'], capsize=5, error_kw={'alpha': 0.7, 'color': 'gray'})

# Ajustar ylim para dados reais do Bahrein
min_time = min(bahrain_2022_q1_times_plot.min(), bahrain_2023_q1_times_plot.min())
max_time = max(bahrain_2022_q1_times_plot.max(), bahrain_2023_q1_times_plot.max())
axes[0].set_ylim(min_time - 1, max_time + 1)
axes[0].set_title("Tempo médio de qualificação Q1 por ano\ncom Intervalo de Confiança 95%", fontsize=11, fontweight='bold', pad=10)
axes[0].set_ylabel("Tempo Médio de Volta (s)", fontsize=10)
axes[0].set_xlabel("Ano", fontsize=10)

# Rótulos nas barras
for i, bar in enumerate(bars):
    yval = bar.get_height()
    error_val = stats_df['yerr'].iloc[i]
    axes[0].text(bar.get_x() + bar.get_width()/2.0, yval + error_val + 0.08,
                 f"{yval:.2f}s", ha='center', va='bottom', fontweight='bold')

# -----------------------------------------------------------------------------
# PASSO 2: DISPERSÃO VERTICAL / JITTER (STRIP PLOT)
# -----------------------------------------------------------------------------
sns.stripplot(
    data=df_melted, x='Ano', y='Tempo_s', hue='Ano', ax=axes[1],
    palette=[Azul, Amarelo], size=9, jitter=0.15,
    linewidth=0.5, alpha = 0.7, edgecolor='black', legend=False
)
axes[1].set_title("Visualização da dispersão dos tempos de volta Q1", fontsize=11, fontweight='bold', pad=10)
axes[1].set_ylabel("Tempo de Volta (s)", fontsize=10)
axes[1].set_xlabel("Ano", fontsize=10)
axes[1].set_ylim(min_time - 1, max_time + 1)

# -----------------------------------------------------------------------------
# PASSO 3: QQ-PLOT EMPÍRICO
# -----------------------------------------------------------------------------
q_2022 = np.sort(df['2022'])
q_2023 = np.sort(df['2023'])

axes[2].scatter(q_2022, q_2023, color='black', s=60, edgecolor='black', linewidth=0.5, zorder=3)

# Diagonal de referência y = x
lims = [min(q_2022.min(), q_2023.min()), max(q_2022.max(), q_2023.max())]
axes[2].plot(lims, lims, 'r--', alpha=0.8, linewidth=1.5, label='Referência y = x (Distribuições Idênticas)')

# Ajustar xlim e ylim para dados reais do Bahrein
axes[2].set_xlim(min_time - margin, max_time + margin)
axes[2].set_ylim(min_time - margin, max_time + margin)

axes[2].set_title("QQ-Plot: Bahrein 2022 vs 2023", fontsize=11, fontweight='bold', pad=10)
axes[2].set_xlabel("Tempo 2022 em segundos (Quantis)", fontsize=10)
axes[2].set_ylabel("Tempo 2023 em segundos (Quantis)", fontsize=10)
axes[2].legend(loc='lower right')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats # Import added here

# Configuração visual
sns.set_theme(style="whitegrid")
plt.rcParams["font.sans-serif"] = "DejaVu Sans"
Azul="#0b1f3b"
Amarelo="#ffc857"

# Fixar semente para reprodutibilidade dos dados de Zandvoort Q1 (20 pilotos)
np.random.seed(42)

# Replace simulated data with real data for Bahrain 2022 (Q1 times)
# zand_2022 is now 'bahrain_2022_q1_times'
bahrain_2022_q1_times = np.sort([
    convert_to_seconds(d['q1']) for d in data_bahrain_2022 if d['q1'] is not None
])

# Ensure both arrays have the same length for proper comparison (if needed), pad with NaN if necessary
# For this QQ plot, we need 20 elements, so let's adjust zand_2023 for now or truncate bahrain_2022_q1_times
# For now, let's proceed assuming zand_2023 will also be real data or adjusted.
# If lengths mismatch due to None values, we need to decide how to handle it.

# Simulação de tempos de volta Q1 para 2023 (mantendo o mesmo número de pontos para o exemplo)
# This will be replaced later if the user provides real 2023 data.
zand_2023 = np.sort(np.random.normal(loc=71.10, scale=0.22, size=len(bahrain_2022_q1_times)))

# DataFrame estruturado
df = pd.DataFrame({
    'Piloto_Rank': range(1, len(bahrain_2022_q1_times) + 1),
    '2022': bahrain_2022_q1_times,
    '2023': zand_2023
})

df_melted = pd.melt(df, id_vars=['Piloto_Rank'], value_vars=['2022', '2023'],
                    var_name='Ano', value_name='Tempo_s')

# Criando o painel de 3 gráficos
fig, axes = plt.subplots(1, 3, figsize=(18, 5.5), dpi=100)

# -----------------------------------------------------------------------------
# PASSO 1: GRÁFICO DE BARRAS (MÉDIA)
# -----------------------------------------------------------------------------
# Calculate statistics for mean and confidence interval
stats_df = df_melted.groupby('Ano')['Tempo_s'].agg(
    mean='mean',
    std='std',
    count='count'
).reset_index()

# Calculate Standard Error of the Mean (SEM)
stats_df['sem'] = stats_df['std'] / np.sqrt(stats_df['count'])

# Calculate 95% Confidence Interval for the mean using t-distribution
# yerr in bar chart expects the positive error value (distance from mean to upper bound)
stats_df['yerr'] = stats_df.apply(
    lambda row: row['mean'] - stats.t.interval(0.95, df=row['count']-1, loc=row['mean'], scale=row['sem'])[0]
    if row['count'] > 1 else 0,
    axis=1
)

bars = axes[0].bar(stats_df['Ano'], stats_df['mean'], color=[Azul, Amarelo],
                   width=0.45, alpha = 0.7, edgecolor='black', yerr=stats_df['yerr'], capsize=5, error_kw={'alpha': 0.7, 'color': 'gray'})
axes[0].set_ylim(69, 94) # Adjusted ylim
axes[0].set_title("Tempo médio de qualificação por ano\ncom Intervalo de Confiança 95%", fontsize=11, fontweight='bold', pad=10)
axes[0].set_ylabel("Tempo Médio de Volta (s)", fontsize=10)
axes[0].set_xlabel("Ano", fontsize=10)

# Rótulos nas barras
for i, bar in enumerate(bars):
    yval = bar.get_height()
    error_val = stats_df['yerr'].iloc[i]
    axes[0].text(bar.get_x() + bar.get_width()/2.0, yval + error_val + 0.08,
                 f"{yval:.2f}s", ha='center', va='bottom', fontweight='bold')

# -----------------------------------------------------------------------------
# PASSO 2: DISPERSÃO VERTICAL / JITTER (STRIP PLOT)
# -----------------------------------------------------------------------------
sns.stripplot(
    data=df_melted, x='Ano', y='Tempo_s', hue='Ano', ax=axes[1],
    palette=[Azul, Amarelo], size=9, jitter=0.15,
    linewidth=0.5, alpha = 0.7, edgecolor='black', legend=False
)
axes[1].set_title("Visualização da dispersão dos tempos de volta", fontsize=11, fontweight='bold', pad=10)
axes[1].set_ylabel("Tempo de Volta (s)", fontsize=10)
axes[1].set_xlabel("Ano", fontsize=10)
axes[1].set_ylim(69, 94) # Adjusted ylim

# -----------------------------------------------------------------------------
# PASSO 3: QQ-PLOT EMPÍRICO
# -----------------------------------------------------------------------------
q_2022 = np.sort(df['2022'])
q_2023 = np.sort(df['2023'])

axes[2].scatter(q_2022, q_2023, color='black', s=60, edgecolor='black', linewidth=0.5, zorder=3)

# Diagonal de referência y = x
lims = [min(q_2022.min(), q_2023.min()), max(q_2022.max(), q_2023.max())] # Adjusted lims to cover both data ranges
axes[2].plot(lims, lims, 'r--', alpha=0.8, linewidth=1.5, label='Referência y = x (Distribuições Idênticas)')
axes[2].set_xlim(90, 94) # Adjusted xlim for Bahrain 2022
axes[2].set_ylim(70, 72.5) # Adjusted ylim for simulated 2023
axes[2].set_title("Passo 3: O QQ-Plot Diagnostica\n(Mudança de inclinação = Redução de variância)", fontsize=11, fontweight='bold', pad=10)
axes[2].set_xlabel("Tempo 2022 em segundos (Quantis)", fontsize=10)
axes[2].set_ylabel("Tempo 2023 em segundos (Quantis)", fontsize=10)
axes[2].legend(loc='lower right')

plt.tight_layout()
plt.show()
```
