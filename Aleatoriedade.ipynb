```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configuração visual limpa
sns.set_theme(style="white")

# Parâmetros
N_pontos = 100
proporcao_x, proporcao_y = 9, 16

# ==========================================
# 1. A ILUSÃO DE ALEATORIEDADE (Jittered Grid)
# ==========================================
# Criando a primeira figura
fig1, ax1 = plt.subplots(figsize=(4.5, 8)) # Ajuste o figsize para uma única figura

# Calculando a grade para acomodar ~1000 pontos na proporção 9:16
# nx * ny = 1000 e ny/nx = 16/9 -> nx ~= 24, ny ~= 42
nx, ny = 8, 16
x_grid = np.linspace(0, proporcao_x, nx)
y_grid = np.linspace(0, proporcao_y, ny)
xv, yv = np.meshgrid(x_grid, y_grid)

# Adicionando um ruído (jitter) menor que o tamanho da célula para evitar sobreposição excessiva
jitter_x = np.random.uniform(-0.4 * (proporcao_x/nx), 0.4 * (proporcao_x/nx), xv.shape)
jitter_y = np.random.uniform(-0.4 * (proporcao_y/ny), 0.4 * (proporcao_y/ny), yv.shape)

x_fake = (xv + jitter_x).flatten()
y_fake = (yv + jitter_y).flatten()

ax1.scatter(x_fake, y_fake, s=15, alpha=0.7, color="#2c7fb8", edgecolors='none')
ax1.set_title("O 'Aleatório' Humano\n(Espaçamento Uniforme)", fontsize=14, pad=15)
ax1.set_xlim(0, proporcao_x)
ax1.set_ylim(0, proporcao_y)
ax1.axis('off') # Remove eixos para focar nos pontos

plt.tight_layout()
plt.show()

# ==========================================
# 2. A ALEATORIEDADE REAL (Distribuição Uniforme)
# ==========================================
# Criando a segunda figura
fig2, ax2 = plt.subplots(figsize=(4.5, 8)) # Ajuste o figsize para uma única figura

# Geração puramente aleatória nos eixos X e Y independentes
x_true = np.random.uniform(0, proporcao_x, N_pontos)
y_true = np.random.uniform(0, proporcao_y, N_pontos)

ax2.scatter(x_true, y_true, s=15, alpha=0.7, color="#d95f02", edgecolors='none')
ax2.set_title("Aleatoriedade Real\n(Distribuição Uniforme)", fontsize=14, pad=15)
ax2.set_xlim(0, proporcao_x)
ax2.set_ylim(0, proporcao_y)
ax2.axis('off')

plt.tight_layout()
plt.show()
```
